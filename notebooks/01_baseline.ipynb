```python
# Locate the existing checkout or clone it in Colab.
import os
import subprocess
from pathlib import Path

repository_url = "https://github.com/geethikagattu/OCSA-Deepfake-Detection.git"
current_path = Path.cwd().resolve()
repo_root = current_path

while repo_root != repo_root.parent and not (repo_root / "requirements.txt").exists():
    repo_root = repo_root.parent

if not (repo_root / "requirements.txt").exists():
    repo_root = Path("/content/OCSA-Deepfake-Detection").resolve()
    if not repo_root.exists():
        subprocess.run(["git", "clone", repository_url, str(repo_root)], check=True)

os.chdir(repo_root)
print("Repository:", repo_root)
```
